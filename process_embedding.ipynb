```python
import basesetup as setup
import pandas as pd

n = 128
embedding_method = 'raster1' # hilbert raster1 raster2

df = pd.read_csv(f'./img_data_{n}.csv')

df_emb = setup.embedding_images_from_dataframe(df, output_path=f'./emb/{n}/{embedding_method}', size=(n,n), embedding_method=embedding_method, verbose=False, override=True)

print(f"Total images transformed: {len(df_emb)}")
print(df_emb.head())

df_emb.to_csv(f'./img_data_{n}_emb_{embedding_method}.csv', index=False)
```
